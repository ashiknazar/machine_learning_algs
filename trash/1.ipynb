# Clustering

## 1. What is Clustering?

**Clustering** is an *unsupervised learning* technique that groups data points so that points in the **same cluster are more similar** to each other than to points in other clusters. There are no labels; the algorithm finds structure on its own.

- **Similarity** is usually measured with a distance, such as Euclidean: $d(p,q)=\sqrt{(x_1-x_2)^2+(y_1-y_2)^2}$
- **Uses:** customer segmentation, image compression, anomaly detection, document grouping

## 3. Clustering Algorithms

| Algorithm | Type | Idea | Example use |
|-----------|------|------|-------------|
| **K-Means** | Centroid-based | Assign points to the nearest of *k* centroids, then update the centroids | Customer segmentation |
| **DBSCAN** | Density-based | Dense regions form clusters, sparse points are noise | Anomaly detection, geo data |
| **Hierarchical** | Connectivity-based | Merge or split clusters step by step (dendrogram) | Gene and taxonomy trees |
| **k-NN** | *Supervised* (not clustering) | Classify a point by its *k* nearest labelled neighbours | Recommendation, classification |

> **Note:** k-NN is a *classification* algorithm, not a clustering one. It is included because it is often compared with K-Means, and it uses the same distance idea.

### 3.1 K-Means (k = 2) – Step by Step

**Algorithm:** (1) choose *k* initial centroids → (2) assign each point to the nearest centroid → (3) recompute each centroid as the mean of its points → (4) repeat 2–3 until assignments stop changing.

**Step 1 – Initial centroids:** C1 = P1 = (1, 1), C2 = P4 = (5, 7)

**Step 2 – Iteration 1, distances**

| Point | d to C1 (1,1) | d to C2 (5,7) | Assigned to |
|-------|---------------|---------------|-------------|
| P1 (1,1) | 0 | $\sqrt{16+36}=7.21$ | C1 |
| P2 (1.5,2) | $\sqrt{0.25+1}=1.12$ | $\sqrt{12.25+25}=6.10$ | C1 |
| P3 (3,4.5) | $\sqrt{4+12.25}=4.03$ | $\sqrt{4+6.25}=3.20$ | C2 |
| P4 (5,7) | 7.21 | 0 | C2 |
| P5 (3.5,5) | $\sqrt{6.25+16}=4.72$ | $\sqrt{2.25+4}=2.50$ | C2 |
| P6 (4.5,5) | $\sqrt{12.25+16}=5.32$ | $\sqrt{0.25+4}=2.06$ | C2 |
| P7 (3.5,4.5) | $\sqrt{6.25+12.25}=4.30$ | $\sqrt{2.25+6.25}=2.92$ | C2 |

Clusters: **C1 = {P1, P2}**, **C2 = {P3, P4, P5, P6, P7}**

**Step 3 – Update centroids**

- C1 = $\left(\frac{1+1.5}{2},\ \frac{1+2}{2}\right) = (1.25,\ 1.5)$
- C2 = $\left(\frac{3+5+3.5+4.5+3.5}{5},\ \frac{4.5+7+5+5+4.5}{5}\right) = (3.9,\ 5.2)$

**Step 4 – Iteration 2, distances**

| Point | d to C1 (1.25,1.5) | d to C2 (3.9,5.2) | Assigned to |
|-------|--------------------|-------------------|-------------|
| P1 | 0.56 | 5.10 | C1 |
| P2 | 0.56 | 4.50 | C1 |
| P3 | 3.47 | 1.14 | C2 |
| P4 | 6.66 | 2.11 | C2 |
| P5 | 4.16 | 0.45 | C2 |
| P6 | 4.78 | 0.63 | C2 |
| P7 | 3.75 | 0.81 | C2 |

Assignments did **not change**, so the algorithm has **converged**.

**Final result:** Cluster 1 = {P1, P2} with centroid (1.25, 1.5). Cluster 2 = {P3, P4, P5, P6, P7} with centroid (3.9, 5.2).

### 3.2 k-NN (k = 3) – Step by Step

**Goal:** classify a new point **Q = (3, 3)** using the clusters above as labels (A = {P1, P2}, B = {P3–P7}).

**Algorithm:** (1) compute the distance from Q to every point → (2) sort ascending → (3) take the *k* nearest → (4) majority vote.

**Step 1 – Distances from Q (3,3)**

| Point | Calculation | Distance | Label |
|-------|-------------|----------|-------|
| P1 (1,1) | $\sqrt{4+4}$ | 2.83 | A |
| P2 (1.5,2) | $\sqrt{2.25+1}$ | 1.80 | A |
| P3 (3,4.5) | $\sqrt{0+2.25}$ | 1.50 | B |
| P4 (5,7) | $\sqrt{4+16}$ | 4.47 | B |
| P5 (3.5,5) | $\sqrt{0.25+4}$ | 2.06 | B |
| P6 (4.5,5) | $\sqrt{2.25+4}$ | 2.50 | B |
| P7 (3.5,4.5) | $\sqrt{0.25+2.25}$ | 1.58 | B |

**Step 2 – Sort:** P3 (1.50) < P7 (1.58) < P2 (1.80) < P5 (2.06) < P6 (2.50) < P1 (2.83) < P4 (4.47)

**Step 3 – Take k = 3 nearest:** P3 (B), P7 (B), P2 (A)

**Step 4 – Vote:** B = 2, A = 1 → **Q belongs to class B**

### 3.3 DBSCAN (ε = 1.5, minPts = 3) – Step by Step

**Definitions**

- **ε (eps):** neighbourhood radius
- **minPts:** minimum points (including itself) within ε for a point to be a **core point**
- **Border point:** within ε of a core point but not a core itself
- **Noise:** neither core nor border

**Algorithm:** (1) for each point, find its ε-neighbours → (2) mark core points → (3) start a cluster from an unvisited core point and expand through all density-reachable points → (4) points not reached are noise.

**Step 1 – ε-neighbourhoods (distance ≤ 1.5)**

| Point | Neighbours within 1.5 (incl. itself) | Count | Type |
|-------|--------------------------------------|-------|------|
| P1 | P1, P2 (d=1.12) | 2 | not core |
| P2 | P2, P1 (d=1.12) | 2 | not core |
| P3 | P3, P7 (0.50), P5 (0.71) | 3 | **core** |
| P4 | P4 only (nearest P6 is 2.06) | 1 | not core |
| P5 | P5, P3 (0.71), P7 (0.50), P6 (1.00) | 4 | **core** |
| P6 | P6, P5 (1.00), P7 (1.12) | 3 | **core** |
| P7 | P7, P3 (0.50), P5 (0.50), P6 (1.12) | 4 | **core** |

(P3–P6 distance is $\sqrt{2.25+0.25}=1.58 > 1.5$, so they are not direct neighbours.)

**Step 2 – Build the cluster:** start at P3 (core) → add P7, P5 → from P5 add P6 → no new points. **Cluster 1 = {P3, P5, P6, P7}**

**Step 3 – Remaining points:** P1 and P2 are not core, and no core point is within 1.5 of them, so they are **noise**. P4 has no neighbours, so it is also **noise**.

**Final result:** 1 cluster {P3, P5, P6, P7}, noise = {P1, P2, P4}.
With **minPts = 2**, {P1, P2} would become a second cluster, which shows how sensitive DBSCAN is to its parameters.

### Comparison Summary

| | K-Means | k-NN | DBSCAN |
|---|---|---|---|
| Learning type | Unsupervised | Supervised | Unsupervised |
| Needs number of clusters? | Yes (k) | No (needs k neighbours + labels) | No (needs ε, minPts) |
| Handles noise/outliers? | Poorly | Moderately | Yes, labels them noise |
| Cluster shape | Spherical | n/a | Arbitrary |
