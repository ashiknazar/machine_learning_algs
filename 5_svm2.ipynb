##### The hyperplane equation

w · x + b = 0

**Classifying a point**: compute w · x + b. If it’s positive, the point is class +1. If it’s negative, the point is class −1.

#### The margin rules

- svm scales **w** and **b** so the closest point on each side give exactly **±1**


The gap between these two edges is:

margin = 2 / ||w||

So maximizing the margin means minimizing ||w||, subject to every point being on the correct side:

yᵢ (w · xᵢ + b) ≥ 1   for all points

The points that sit exactly on the edges, where yᵢ(w · xᵢ + b) = 1, are the support vectors. Only they decide the line. You could delete all other points and get the same result.

- $w.x+b=0$

- Why w · x + b = 0 is the hyperplane
   - In 2D, any line can be written $ax + cy + d = 0$
   - Rename the coefficients: w = (w₁, w₂) and b
   - w₁x₁ + w₂x₂ + b = 0
- In 3D it’s a plane,in higher dimensions it’s a hyperplane.


- Why **w** is perpendicular to the line: take two points A and B that both lie on the line.
