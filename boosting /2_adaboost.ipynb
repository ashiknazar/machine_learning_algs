### AdaBoost: Amount of Say (Learner Weight)

$$
\alpha = \frac{1}{2}\log\left(\frac{1-\text{total\_error}}{\text{total\_error}}\right)
$$

- Laplace smoothing (epsilon adjustment) to avoid division by zero and logarithm errors when the total error is 0 or 1.


$$
\alpha = \frac{1}{2}\log\left(\frac{1-\text{total\_error}+\epsilon}{\text{total\_error}+\epsilon}\right)
$$

 - when prediction is just a random assumption or half correct- total error=0.5 , then amount of say will be 0
 - when stump is terrible total_error=1,amount of say will be large -ve

### AdaBoost: Weight Update for Misclassified Samples

$$
w_i^{(t+1)} = w_i^{(t)} \times e^{\alpha_t}
$$

Misclassified samples have their weights increased by multiplying them by $e^{\alpha_t}$, making them more important for the next weak learner.

$ {\alpha}=amountofsay$

![](wrong.png)   

- when amount of say is relatively large (ie the last stump did a good job classifying samples)
- scale previous weight with a large number
- new sample weight will be much larger than old one 

- when amount of say is relatevly low the last stump did not do a very good job 
- we scale previous sample weight with a relatively small number
- new sample weight will only be a little larger than old one
