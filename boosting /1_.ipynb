| Algorithm | Main Concept |
|---|---|
| AdaBoost | Increases weights of incorrectly classified samples. |
| Gradient Boosting | Trains new models to minimize previous prediction errors using gradients. |
| XGBoost | Optimized gradient boosting with regularization and computational improvements. |
| LightGBM | Histogram-based gradient boosting designed for speed and large datasets. |
| CatBoost | Gradient boosting designed to handle categorical features effectively. |
| HistGradientBoosting | Uses histogram-based feature binning for efficient gradient boosting. |
